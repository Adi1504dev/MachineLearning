```python
#Ann Regression
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
ann_model=Sequential([
    Dense(64,activation='relu',input_shape=(X_train.shape[1],)),
    Dense(32,activation='relu'),
    Dense(1,activation='linear')#linear is default
])

ann_model.compile(optimizer='adam',loss='mean_absolute_error',metrics=['mae'])
ann_model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
#setup Tensor Board
log_dir="regression_log/fit/"+datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
```

```python
ts_callback=TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
early_stop_callback=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
ann_model.fit(X_train,y_train,
              validation_data=(X_test,y_test),
              epochs=100,
              callbacks=[ts_callback,early_stop_callback])
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
import pickle
from sklearn.preprocessing import StandardScaler,LabelEncoder
```

```python
## Load the dataset
data=pd.read_csv("Churn_Modelling.csv")
```

```python
data.head()
```

```python
## Preprocess the data
## Drop irrelevent features
data=data.drop(['RowNumber',"CustomerId","Surname"],axis=1)
data.head()
```

```python
# Encode categorical variable
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
labelencoder_gender=LabelEncoder()
OneHotEncoder_geo=OneHotEncoder()
data['Gender']=labelencoder_gender.fit_transform(data['Gender'])
df_encoded_geo=OneHotEncoder_geo.fit_transform(data[['Geography']])
df_encoded_geo=df_encoded_geo.toarray()# To array needs to be done to convert it to DF
```

```python
OneHotEncoder_geo.get_feature_names_out(['Geography'])
```

```python
geo_df=pd.DataFrame(df_encoded_geo,columns=OneHotEncoder_geo.get_feature_names_out(['Geography']))
```

```python
data=pd.concat([data.drop(["Geography"],axis=1),geo_df],axis=1)
```

```python
with open('labelencoder_gender.pkl','wb') as file:
    pickle.dump(labelencoder_gender,file)
```

```python
with open('OneHotEncoder_geo.pkl','wb') as file:
    pickle.dump(OneHotEncoder_geo,file)
```

```python
from sklearn.model_selection import train_test_split

X=data.drop(['Exited'],axis=1)
y=data['Exited']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)

#Scale Trasformation
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
with open('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

```python
import tensorflow as TF
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime

```

```python
#Build ANN model
model=Sequential(
    [
    Dense(64,activation='relu',input_shape=(X_train.shape[1],)),# HL1 connected with input Layer
    Dense(32,activation='relu'),
    Dense(1,activation='sigmoid')
    ]
)
```

```python
model.summary()
```

```python
#We can setup loss and optimizers as per below as well
optimizer=TF.keras.optimizers.Adam(learning_rate=0.01)
loss=TF.keras.losses.BinaryCrossentropy()
```

```python
#compile the model
model.compile(optimizer=optimizer,loss="binary_crossentropy",metrics=['accuracy'])
```

```python
#compile the model
#If u set optimzer in this way u can not give leaning rate
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=['accuracy'])
```

```python
## Setup Tensor Board
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
log_dir="logs/fit/"+datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensor_callback=TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
# Set up Early Stopping
early_stopping_callback= EarlyStopping(monitor='val_loss',patience=5, restore_best_weights=True)
```

```python
# Train the model
history=model.fit(
    X_train,y_train,validation_data=(X_test,y_test),epochs=100,
    callbacks=[tensor_callback,early_stopping_callback]
)
```

```python
model.save('model.h5')
```
